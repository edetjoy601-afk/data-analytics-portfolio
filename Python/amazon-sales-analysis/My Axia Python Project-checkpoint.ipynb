```python
# Exploratory Data Analysis of Amazon Sales Dataset
# Final Python Project – Axia Africa Data Analysis Program

#Importing Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Loading of dataset
df=pd.read_csv("amazon_sales_dataset.csv")
```

```python
#To display the first and last few rows
df.head()      # First 5 rows
df.tail()      # Last 5 rows
```

```python
df.shape       # Rows and columns
df.columns     # Column names
df.info()      # Data types
```

```python
#Checking for missing values
print("Missing values per column:")
print(df.isnull().sum())

#emove duplicates
df = df.drop_duplicates()

# Convert Order Date to datetime
if 'order_date' in df.columns:
    df['order_date'] = pd.to_datetime(df['order_date'])
```

```python
#1 To calculate total sales, total profit and total quantity sold
#Total sales
Total_Revenue= df['total_revenue'].sum()
print(Total_Revenue)
#Total profit
df['Cost'] = df['price'] * df['quantity_sold']
df['Profit'] = df['Cost'] - df['total_revenue']

Total_Profit = df['Profit'].sum()

print(Total_Profit)

#Total Quantity sold
Total_Quantity_Sold= df['quantity_sold'].sum()
print(Total_Quantity_Sold)

#The analysis shows that the business generated significant total revenue from all transactions, with a large volume of products sold. 
#The calculated profit, based on the assumed cost structure, indicates that the business is generating returns, although profitability may be influenced by pricing and discount strategies.
```

```python
#2 To determine sales trend overtime i.e how sales change overtime(monthly)
monthly_sales = df.groupby(df['order_date'].dt.to_period('M'))['total_revenue'].sum()

monthly_sales = monthly_sales.sort_index()

plt.figure()
monthly_sales.plot(kind='line', marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()

#The monthly sales trend shows fluctuations in revenue over time, indicating variations in customer demand across different periods. 
#Some months recorded higher sales peaks, while others experienced lower performance. 
#This pattern may suggest seasonal trends, promotional impacts, or changing customer purchasing behavior over time.
```

```python
#3 To determine which product categories generate the highest sales and profit(Category Performance)
category_performance = df.groupby('product_category').agg({
    'total_revenue': 'sum',
    'Profit': 'sum'
}).sort_values(by='total_revenue', ascending=False)

print(category_performance)
top_sale_category = category_performance.head(1)
print(top_sale_category)

# Create a bar chart
sns.barplot(x='total_revenue', y='product_category', data=df)
plt.title("category_performance")
plt.show()

sns.barplot(x='Profit', y='product_category', data=df)
plt.title("category_performance")
plt.show()

#The analysis reveals differences in sales and profit across product categories. 
#Some categories generate higher revenue compared to others, indicating stronger customer demand.
#However, high sales do not always correspond to the highest profit, suggesting that pricing, discounts, or cost assumptions may affect profitability across categories.
```

```python
#4 To determine the top 5 sub categories by sales or profit (Sub-category Analysis)
#Subcategory-level analysis could not be performed due to the absence of a subcategory column in the dataset. This analysis will be done with product categories
sub_category_analysis = df.groupby('product_category').agg({
    'total_revenue': 'sum',
    'Profit': 'sum'
}).sort_values(by='total_revenue', ascending=False)

print(sub_category_analysis)

# Create a bar chart
sns.barplot(x='total_revenue', y='product_category', data=df)
plt.title("sub_category_analysis")
plt.show()

sns.barplot(x='Profit', y='product_category', data=df)
plt.title("sub_category_analysis")
plt.show()

#Due to the absence of a subcategory column in the dataset, product category was used as an alternative for analysis.
#The results show that a few categories contribute more significantly to overall sales and profit, highlighting areas of strong product performance.
```

```python
#5 To check for regonal performance, which region performs best in terms of sales and profit 
region_performance = df.groupby('customer_region').agg({
    'total_revenue': 'sum',
    'Profit': 'sum'
}).sort_values(by='total_revenue', ascending=False)

print(region_performance)
best_performing_region = region_performance.head(1)
print(best_performing_region)

# Create a column chart
sns.barplot(x='customer_region', y='total_revenue', data=df)
plt.title("Region_Performance")
plt.show()

sns.barplot(x='customer_region', y='Profit', data=df)
plt.title("Region_Performance")
plt.show()

#The analysis shows that some regions outperform others in terms of total sales and profit. 
#The top-performing region (Middle East) contributes the highest revenue, indicating stronger market demand or better customer engagement in that location. 
#This suggests that regional strategies may play a key role in business performance.
```

```python
#6 To determine if there are categories or subcategories with high sales but low profit
category_performance = df.groupby('product_category').agg({
    'total_revenue': 'sum',
    'Profit': 'sum'
}).sort_values(by='total_revenue', ascending=False)

print(category_performance)

#The comparison between sales and profit across product categories shows that high sales do not always lead to high profitability. 
#Some categories generate strong revenue but relatively lower profit, indicating the possible impact of discounting, pricing strategies, or cost assumptions on overall profitability.
```

```python
#7 Customer Purchasing Behaviour
#To determine the distribution of order quantities
import matplotlib.pyplot as plt

plt.figure()
plt.hist(df['quantity_sold'], bins=10)
plt.title('Distribution of Quantity Sold')
plt.xlabel('quantity_sold')
plt.ylabel('Frequency')
plt.show()

#To determine if customers tend to buy in small or large quantities
print(df['quantity_sold'].describe())

#The distribution of order quantities shows that customers consistently purchase between 1 and 5 items per order, with an average of approximately 3 items. 
#This indicates that customers tend to buy in small quantities, suggesting a retail-oriented purchasing pattern rather than bulk buying behavior.
```

```python
#8 Relationship Analysis: This is to determine the relationship between sales, quantity and profit 
#Using correlation analysis and visualization.
# +1 means strong positive relationship
# 0 means no relationsihp
# -1 means strong negative relationship

correlation = df[['total_revenue', 'quantity_sold', 'Profit']].corr()

print(correlation)

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure()
sns.heatmap(correlation, annot=True)
plt.title('Correlation Between Sales, Quantity, and Profit')
plt.show()

#The correlation analysis shows a positive relationship between quantity sold and total revenue, indicating that higher quantities contribute to increased sales. 
#There is also a moderate relationship between profit and both sales and quantity, suggesting that profitability is influenced by multiple factors beyond just sales volume.
```
